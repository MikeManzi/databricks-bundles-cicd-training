# Verification Notebook

This notebook creates a test table with sample data for CI/CD verification.

```python
# Get parameters
catalog = dbutils.widgets.get("catalog")
schema = dbutils.widgets.get("schema")

print(f"Using catalog: {catalog}")
print(f"Using schema: {schema}")

# Set catalog and schema
spark.sql(f"USE CATALOG {catalog}")
spark.sql(f"CREATE SCHEMA IF NOT EXISTS {schema}")
spark.sql(f"USE SCHEMA {schema}")
```

```python
# Create a test table with sample data
from datetime import datetime
from pyspark.sql import Row

# Create sample data
data = [
    Row(id=1, name="Test Record 1", timestamp=datetime.now()),
    Row(id=2, name="Test Record 2", timestamp=datetime.now()),
    Row(id=3, name="Test Record 3", timestamp=datetime.now())
]

df = spark.createDataFrame(data)

# Write to table
table_name = "cicd_verification_table"
df.write.mode("overwrite").saveAsTable(table_name)

print(f"Created table: {catalog}.{schema}.{table_name}")
print(f"Row count: {df.count()}")

# Display the table
display(spark.table(table_name))
```
